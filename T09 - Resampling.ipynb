### 1.1. Derivación de (5.6)
- Usar propiedades básicas de la varianza.
- Utilizar cálculo de una variable.
- Probar que la expresión dada en (5.6) minimiza  $\mathrm{Var}\!\left( \alpha X + (1-\alpha)Y \right)$

# Demostración de la ecuación (5.6)

Queremos demostrar que la cantidad

$\displaystyle 
\lambda^\=\frac{\sigma_Y^2-\sigma_{XY}}
{\sigma_X^2+\sigma_Y^2-2\sigma_{XY}}
$

minimiza la varianza de la combinación lineal

$\mathrm{Var}(\lambda X + (1-\lambda)Y)$,

donde $\sigma_X^2 = \mathrm{Var}(X)$, $\sigma_Y^2 = \mathrm{Var}(Y)$ y $\sigma_{XY} = \mathrm{Cov}(X,Y)$.

---

## 1. Expansión de la varianza

Usamos la identidad:

$\mathrm{Var}(aX + bY) = a^2\sigma_X^2 + b^2\sigma_Y^2 + 2ab\sigma_{XY}$.

En este caso $a=\lambda$ y $b=1-\lambda$:

$\mathrm{Var}(\lambda X + (1-\lambda)Y)
= \lambda^2\sigma_X^2 + (1-\lambda)^2\sigma_Y^2 + 2\lambda(1-\lambda)\sigma_{XY}.
$

Expandimos:

$\mathrm{Var}(\lambda) =
\lambda^2\sigma_X^2
+ (1 - 2\lambda + \lambda^2)\sigma_Y^2
+ 2\lambda\sigma_{XY}
- 2\lambda^2\sigma_{XY}.
$

Agrupamos términos:

$\mathrm{Var}(\lambda)
= \lambda^2(\sigma_X^2 + \sigma_Y^2 - 2\sigma_{XY})
+ \lambda\, [-2\sigma_Y^2 + 2\sigma_{XY}]
+ \sigma_Y^2.
$

Sea:

$A = \sigma_X^2 + \sigma_Y^2 - 2\sigma_{XY}$  
$B = 2(\sigma_{XY} - \sigma_Y^2)$  
$C = \sigma_Y^2$

Entonces:

$\mathrm{Var}(\lambda) = A\lambda^2 + B\lambda + C.$

---

## 2. Derivada respecto a $\lambda$

$\mathrm{Var}'(\lambda) = 2A\lambda + B.$

---

## 3. Hallando el mínimo

Igualamos la derivada a cero:

$2A\lambda + B = 0$

$\lambda A = -\frac{B}{2}.$

Sustituimos $B = 2(\sigma_{XY}-\sigma_Y^2)$:

$\lambda A = \sigma_Y^2 - \sigma_{XY}.$

Entonces:

$\lambda^\* =
\frac{\sigma_Y^2 - \sigma_{XY}}{A}.
$

Como $A = \sigma_X^2 + \sigma_Y^2 - 2\sigma_{XY}$:

$\boxed{
\lambda^\* =
\frac{\sigma_Y^2 - \sigma_{XY}}
{\sigma_X^2 + \sigma_Y^2 - 2\sigma_{XY}}
}
$

que es exactamente la ecuación (5.6).

---

## 4. Verificación de que es un mínimo

La segunda derivada es:

$\mathrm{Var}''(\lambda) = 2A.$

Notemos que:

$A = \sigma_X^2 + \sigma_Y^2 - 2\sigma_{XY}
= \mathrm{Var}(X - Y) \ge 0.$

En casos no degenerados es estrictamente positiva, así que:

$\mathrm{Var}''(\lambda) > 0,$

confirmando que $\lambda^\*$ es un **mínimo global**.

---

$\boxed{
\lambda^\* =
\frac{\sigma_Y^2 - \sigma_{XY}}
{\sigma_X^2 + \sigma_Y^2 - 2\sigma_{XY}}
}
$

minimiza la varianza de la inversión

$\mathrm{Var}(\lambda X + (1-\lambda)Y).$

### 2.1. Probabilidad de inclusión de una observación en un bootstrap sample

#### (a)
- **Calcular la probabilidad de que la **primera** observación bootstrap **no sea** la observación \(j\)-ésima del conjunto original.**
- **Justificar la respuesta.**

Cada observación del bootstrap se obtiene eligiendo **uniformemente** entre las $n$ observaciones originales.

La probabilidad de seleccionar la observación $j$ en cualquier extracción es:

$\displaystyle P(\text{seleccionar } j)=\frac{1}{n}$.

Por lo tanto, la probabilidad de **no** seleccionar la observación $j$ en la primera posición del bootstrap es:

$\displaystyle P(\text{no seleccionar } j)=1-\frac{1}{n}.$


#### (b)
- **Calcular la probabilidad de que la **segunda** observación bootstrap **no sea** la observación \(j\)-ésima.**

Las extracciones del bootstrap son **independientes y con reemplazo**, así que la probabilidad de que la segunda observación **no** sea $j$ es la misma que en la primera:

$\displaystyle P(\text{no seleccionar } j)=1-\frac{1}{n}.$


#### (c)
- **Argumentar que la probabilidad de que la observación \(j\)-ésima **no aparezca** en el bootstrap sample es  $ (1 - 1/n)^n $**

En total se realizan $n$ selecciones para construir la muestra bootstrap.

La probabilidad de que **ninguna** de las $n$ selecciones elija la observación $j$ es:

$\displaystyle
P(\text{el $j$-ésimo no aparece en el bootstrap})
= \left(1 - \frac{1}{n}\right)^n.
$

Esto se debe a la independencia entre selecciones con reemplazo.


#### (d)
- **Para \(n = 5\), calcular la probabilidad de que la observación \(j\)-ésima **sí esté** en el bootstrap sample.**

Para $n=5$, la probabilidad de que la observación $j$ **sí** esté en el bootstrap es:

$\displaystyle
P(\text{aparece}) = 1 - \left(1 - \frac{1}{5}\right)^5
= 1 - \left(\frac{4}{5}\right)^5.
$

Numéricamente:

$\displaystyle
1 - \left(\frac{4}{5}\right)^5 = 1 - 0.32768 = 0.67232.
$

#### (e)
- **Para \(n = 100\), calcular la misma probabilidad.**

Para $n=100$:

$\displaystyle
P(\text{aparece}) = 1 - \left(1 - \frac{1}{100}\right)^{100}.
$

Sabemos que:

$\left(1 - \frac{1}{100}\right)^{100} \approx e^{-1} \approx 0.3679.$

Entonces:

$\displaystyle
1 - 0.3679 \approx 0.6321.
$


#### (f)
- **Para \(n = 10,000\), calcular la misma probabilidad.**

Para $n=10,000$:

$\displaystyle
P(\text{aparece}) = 1 - \left(1 - \frac{1}{10000}\right)^{10000}.
$

Al igual que antes:

$\left(1 - \frac{1}{10000}\right)^{10000} \approx e^{-1}.$

Entonces la probabilidad sigue siendo:

$\displaystyle
1 - e^{-1} \approx 0.6321.
$

#### (g)
- **Crear un **gráfico** para \(n = 1, 2, .............., 100000\) mostrando la probabilidad de que la observación \(j\)-ésima esté en el bootstrap sample.**
- **Comentar la tendencia observada.**

```python
import numpy as np
import matplotlib.pyplot as plt

n_vals = np.arange(1, 100001)
prob_in = 1 - (1 - 1/n_vals)**n_vals

plt.figure(figsize=(8,5))
plt.plot(n_vals, prob_in)
plt.xscale("log")
plt.xlabel("n (escala log)")
plt.ylabel("Probabilidad de que j esté en el bootstrap")
plt.title("Probabilidad de inclusión en función de n")
plt.grid(True)
plt.show()
```

La probabilidad converge rápidamente hacia:

$\displaystyle 1 - e^{-1} \approx 0.6321.$

Incluso para valores moderados de $n$ (como 50 o 100), la curva ya está muy cerca de $0.6321$.
Esto explica por qué aproximadamente el 63.2%, en promedio, de las observaciones aparecen en cualquier muestra bootstrap.

#### (h)
- **Realizar una simulación numérica para \(n = 100\) y \(j = 4\):**
  - **Crear arreglo con `np.empty()`.**
  - **Generar 10,000 bootstrap samples.**
  - **Registrar si la observación 4 aparece en cada bootstrap sample.**
  - **Calcular la proporción con `np.mean(store)`.**

```python
rng = np.random.default_rng(10)
store = np.empty(10000)

for i in range(10000):
    sample = rng.integers(0, 100, size=100)  
    store[i] = (4 in sample)

np.mean(store)
```

El resultado numérico esta muy cerca de:

$\displaystyle 1 - e^{-1} \approx 0.6321.$

En la práctica, se obtiene un valor alrededor de 0.63, confirmando la teoría.

Esto muestra que en un bootstrap de tamaño $n=100$, la observación $j$ (aquí $j=4$) aparece en la muestra bootstrap aproximadamente el 63% de las veces, lo que coincide con la probabilidad límite teórica.

  <h3> 3. k-fold Cross-Validation</h3>
</div>


### 3.1. Implementación
- **Explicar cómo se lleva a cabo k-fold cross-validation.**

1. Se divide el conjunto de datos en $k$ subconjuntos (folds) aproximadamente del mismo tamaño.
2. Por cada fold $i=1,2,\dots,k$:
   - Se toma el fold $i$ como **conjunto de validación**.
   - Se toman los $k-1$ folds restantes como **conjunto de entrenamiento**.
   - Se entrena el modelo con los folds de entrenamiento y se calcula la **métrica de error** sobre el fold de validación.
3. Se repite el proceso hasta que cada fold haya sido usado exactamente una vez como conjunto de validación.
4. Finalmente, se promedian las métricas de error obtenidas en cada iteración para obtener la estimación de error global del modelo.

> Nota: Este procedimiento permite usar todo el conjunto de datos tanto para entrenamiento como para validación sin superponer observaciones en la validación.


### 3.2. Comparación

#### (a) Ventajas y desventajas respecto al validation set approach.

**Ventajas:**
- Reduce la **variabilidad** de la estimación del error, ya que se promedian los resultados de $k$ folds.
- Usa **todas las observaciones** para entrenamiento y validación (cada observación se usa para validación una vez), aprovechando mejor los datos.

**Desventajas:**
- Más **computacionalmente costoso**, ya que el modelo se entrena $k$ veces, mientras que el validation set approach solo requiere entrenar una vez.


#### (b) Ventajas y desventajas respecto a LOOCV.

**Ventajas de k-fold sobre LOOCV:**
- Menor **costo computacional**: se entrena $k$ veces en lugar de $n$ veces (donde $n$ es el número de observaciones).
- Menor **varianza de la estimación del error**, ya que cada fold contiene varias observaciones y el promedio es más estable.

**Desventajas:**
- La estimación de error puede ser **menos precisa** que LOOCV si $k$ es pequeño, porque cada modelo se entrena con menos datos que LOOCV (que usa $n-1$ observaciones para entrenamiento en cada iteración).
- Puede ser sensible a la **elección de $k$**, aunque comúnmente se usa $k=5$ o $k=10$.


<div style="background-color: #cce5ff; padding: 15px; border-radius: 8px;">
  <h3> 4. Estimación de la desviación estándar de una predicción</h3>
</div>


### 4.1. Descripción del procedimiento
- **Describir detalladamente cómo estimar la desviación estándar de una predicción $\hat{Y}$ obtenida para cierto valor de \(X\), usando un método de aprendizaje estadístico.**

Supongamos que hemos entrenado un modelo de aprendizaje estadístico para predecir la respuesta $Y$ a partir de un predictor $X$. Para estimar la **desviación estándar** de nuestra predicción en un valor específico $X=x_0$, podemos proceder de la siguiente manera:

1. **Bootstrap del conjunto de datos**:
   - Generamos $B$ muestras bootstrap a partir del conjunto de datos original. Cada muestra se obtiene seleccionando $n$ observaciones **con reemplazo** del conjunto de datos original de tamaño $n$.
   
2. **Entrenamiento del modelo en cada muestra bootstrap**:
   - Para cada muestra bootstrap $b=1,2,\dots,B$, entrenamos el modelo de aprendizaje estadístico y obtenemos la predicción correspondiente para $x_0$, denotada como $\hat{Y}^{(b)}$.

3. **Colección de predicciones**:
   - Después de entrenar el modelo en las $B$ muestras, obtenemos un conjunto de predicciones:
   $$
   \{\hat{Y}^{(1)}, \hat{Y}^{(2)}, \dots, \hat{Y}^{(B)}\}.
   $$

4. **Cálculo de la desviación estándar empírica**:
   - La estimación de la desviación estándar de la predicción en $x_0$ se obtiene como la desviación estándar muestral de las predicciones bootstrap:
   $$
   \widehat{\text{SD}}(\hat{Y}(x_0)) = \sqrt{\frac{1}{B-1} \sum_{b=1}^{B} \left(\hat{Y}^{(b)} - \bar{\hat{Y}}\right)^2},
   $$
   donde
   $$
   \bar{\hat{Y}} = \frac{1}{B}\sum_{b=1}^{B} \hat{Y}^{(b)}
   $$
   es el promedio de las predicciones bootstrap.

5. **Interpretación**:
   - Esta desviación estándar nos da una medida de **incertidumbre asociada a la predicción** en $x_0$, teniendo en cuenta la variabilidad en los datos de entrenamiento.
   - Cuanto mayor sea la desviación estándar, menos confiable es la predicción para ese valor específico de $X$.

> Nota: Este procedimiento se puede aplicar a cualquier método de aprendizaje estadístico, incluso aquellos para los que no existen fórmulas analíticas para la varianza de la predicción. El bootstrap permite estimar empíricamente la variabilidad de la predicción.

### 5.2. Enfoque de validation set para estimar el test error

#### (a)
- **Ajustar el modelo logístico usando `income` y `balance`.**

```python
import pandas as pd
import statsmodels.api as sm
from sklearn.model_selection import train_test_split

df = pd.read_csv("Default.csv")

# Convertimos la variable 'default' a binaria
df['default_bin'] = df['default'].map({'Yes': 1, 'No': 0})

# Variables predictoras
X = df[['income', 'balance']]
X = sm.add_constant(X)  # Añadir intercepto
y = df['default_bin']

# Ajustamos el modelo de regresión logística
logit_model = sm.Logit(y, X).fit()
logit_model.summary()
```

#### (b) Pasos para estimar el test error:
1. **Dividir la muestra en **training set** y **validation set**.
2. **Ajustar el modelo logístico usando solo las observaciones del training.**
3. **Obtener predicciones en el validation set:**
   - **Calcular la probabilidad posterior de default.**
   - **Clasificar como *default* si la probabilidad es $> 0.5$.**
4. **Calcular el validation set error como la fracción de observaciones mal clasificadas.**

```python
### i. División en conjunto de entrenamiento y validación

# Semilla aleatoria
random_state = 42

X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.5, random_state=random_state
)

### ii. Ajuste del modelo usando solo los datos de entrenamiento
logit_train = sm.Logit(y_train, X_train).fit()

### iii. Predicción de la probabilidad de default en el conjunto de validación
y_pred_prob = logit_train.predict(X_val)
y_pred_class = (y_pred_prob > 0.5).astype(int)

### iv. Cálculo del error en el conjunto de validación
validation_error = (y_pred_class != y_val).mean()
validation_error
```

#### (c)
- **Repetir el proceso anterior **tres veces**, usando tres particiones distintas.**
- **Comentar los resultados.**

```python
## (c) Repetición del proceso tres veces con diferentes divisiones
errors = []
for seed in [1, 42, 100]:
    X_train, X_val, y_train, y_val = train_test_split(
        X, y, test_size=0.5, random_state=seed
    )
    logit_train = sm.Logit(y_train, X_train).fit(disp=False)
    y_pred_class = (logit_train.predict(X_val) > 0.5).astype(int)
    error = (y_pred_class != y_val).mean()
    errors.append(error)

errors
```

Los errores de validación pueden variar ligeramente según la división aleatoria de los datos, pero suelen ser consistentes si el tamaño de la muestra es grande.

### 6.2. Bootstrap de coeficientes

#### (b)
- **Crear una función `boot_fn()` que:**
  - **Reciba el data set y un índice.**
  - **Devuelva los coeficientes estimados para `income` y `balance`.**

```python
def boot_fn(data, index):
    """
    data: DataFrame con los datos
    index: lista o array de índices de las observaciones a usar
    Devuelve: coeficientes de income y balance en regresión logística múltiple
    """
    df_boot = data.iloc[index]
    X_boot = df_boot[['income', 'balance']]
    X_boot = sm.add_constant(X_boot)
    y_boot = df_boot['default_bin']
    
    model = sm.GLM(y_boot, X_boot, family=sm.families.Binomial()).fit()
    return model.params[['income', 'balance']]
```

#### (c)
- **Usar `boot_fn()` para estimar los errores estándar de ambos coeficientes mediante bootstrap.**

```python
np.random.seed(42)
B = 1000  # número de muestras bootstrap
coefs = []

n = len(df)

for b in range(B):
    sample_indices = np.random.choice(n, size=n, replace=True)
    coef_b = boot_fn(df, sample_indices)
    coefs.append(coef_b.values)

coefs = np.array(coefs)
se_bootstrap = coefs.std(axis=0)
se_bootstrap
```

### 7.1. Modelo base

#### (a)
- **Ajustar regresión logística prediciendo `Direction` usando `Lag1` y `Lag2`.**

```python
import statsmodels.api as sm

# Cargar el dataset Weekly desde ISLR (paquete R) vía statsmodels
weekly = sm.datasets.get_rdataset("Weekly", package="ISLR").data

# Convertir Direction a binaria: Up=1, Down=0
weekly['Direction_bin'] = weekly['Direction'].map({'Up': 1, 'Down': 0})

# Variables predictoras
X = weekly[['Lag1', 'Lag2']]
X = sm.add_constant(X)
y = weekly['Direction_bin']

# Ajustar modelo de regresión logística
logit_model = sm.GLM(y, X, family=sm.families.Binomial()).fit()
logit_model.summary()
```

#### (b)
- **Ajustar nuevamente el modelo usando **todas salvo la primera observación**.**

```python
X_train = X.iloc[1:]  # todas menos la primera
y_train = y.iloc[1:]

logit_model_leave1 = sm.GLM(y_train, X_train, family=sm.families.Binomial()).fit()
```

#### (c)
- **Usar ese modelo para predecir la dirección de la primera observación:**
  - **Predecir *Up* si**
    $P(\text{Direction} = \text{"Up"} \mid \text{Lag1}, \text{Lag2}) > 0.5$.
- **Determinar si fue clasificada correctamente.**

```python
# Primera observación
X_first = X.iloc[0:1]

# Probabilidad de que la dirección sea Up
prob_up = logit_model_leave1.predict(X_first)[0]

# Clasificación
pred_class = 1 if prob_up > 0.5 else 0

# Comparar con la observación real
actual_class = y.iloc[0]
correct_classification = (pred_class == actual_class)
pred_class, actual_class, correct_classification
```

No fue clasificada correctamente

### 7.2. LOOCV con un for loop

#### (d)
- **Crear un loop para $i = 1, \ldots, n$ que:**
  1. **Ajuste el modelo usando todas menos la observación $i$.**
  2. **Calcule la probabilidad posterior de que el mercado suba para el caso $i$.**
  3. **Prediga la dirección usando el criterio de $0.5$.**
  4. **Registre un $1$ si hay error y un $0$ si no lo hay.**

```python
# Variables predictoras y respuesta como arrays
X_full = weekly[['Lag1', 'Lag2']]
X_full = sm.add_constant(X_full).values  # convertimos a numpy array
y_full = weekly['Direction'].map({'Up': 1, 'Down': 0}).values

n = len(y_full)
errors = np.zeros(n)

for i in range(n):
    # Excluir la i-ésima observación usando np.delete
    X_train = np.delete(X_full, i, axis=0)
    y_train = np.delete(y_full, i, axis=0)
    
    # Ajustar modelo
    model = sm.GLM(y_train, X_train, family=sm.families.Binomial()).fit()
    
    # Observación de test
    X_test = X_full[i:i+1]
    y_test = y_full[i]
    
    # Predecir
    prob_up = model.predict(X_test)[0]
    pred_class = 1 if prob_up > 0.5 else 0
    
    # Guardar error
    errors[i] = 1 if pred_class != y_test else 0

loocv_error = errors.mean()
print("LOOCV error:", loocv_error)
```

#### (e)
- **Calcular el test error LOOCV tomando el promedio de los $n$ valores.**
- **Comentar los resultados.**

```python
loocv_error = errors.mean()
loocv_error
```

Esto indica que aproximadamente el 45 % de las observaciones fueron clasificadas incorrectamente cuando se aplica LOOCV.

### 8.3. LOOCV para modelos polinomiales

#### (c)
- **Usar LOOCV con mínimos cuadrados para ajustar:**
  1. $Y = \beta_0 + \beta_1 X + \varepsilon$
  2. $Y = \beta_0 + \beta_1 X + \beta_2 X^2 + \varepsilon$
  3. $Y = \beta_0 + \beta_1 X + \beta_2 X^2 + \beta_3 X^3 + \varepsilon$
  4. $Y = \beta_0 + \beta_1 X + \beta_2 X^2 + \beta_3 X^3 + \beta_4 X^4 + \varepsilon$

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import LeaveOneOut
from sklearn.metrics import mean_squared_error

# Crear matriz de predictores para cada modelo
X = x.reshape(-1, 1)
degrees = [1, 2, 3, 4]
loocv_errors = []

loo = LeaveOneOut()

for d in degrees:
    poly = PolynomialFeatures(degree=d, include_bias=True)
    X_poly = poly.fit_transform(X)
    
    errors = []
    for train_idx, test_idx in loo.split(X_poly):
        model = LinearRegression().fit(X_poly[train_idx], y[train_idx])
        y_pred = model.predict(X_poly[test_idx])
        errors.append((y[test_idx] - y_pred)**2)
        
    loocv_errors.append(np.mean(errors))

loocv_errors
```

### 9.1. Media de `medv`

#### (a)
- **Estimar la media poblacional $\hat{\mu}$ de `medv`.**

```python
# Cargar dataset Boston desde ISLR
boston = sm.datasets.get_rdataset("Boston", package="MASS").data

# Estimar la media de medv
mu_hat = boston['medv'].mean()
mu_hat
```

#### (b)
- **Estimar el error estándar de $\hat{\mu}$ usando** 
  $\text{SE}(\hat{\mu}) = \dfrac{s}{\sqrt{n}}$.
- **Interpretar el resultado.**

```python
n = len(boston)
se_mu = boston['medv'].std(ddof=1) / np.sqrt(n)
se_mu
```

La media tiende a desviarce un +- 0.408

### 9.2. Bootstrap para la media

#### (c)
- **Estimar el error estándar de $\hat{\mu}$ usando bootstrap.**
- **Comparar con (b).**

```python
rng = np.random.default_rng(1)
B = 10000
bootstrap_means = np.empty(B)

for b in range(B):
    sample = rng.choice(boston['medv'], size=n, replace=True)
    bootstrap_means[b] = sample.mean()

se_mu_boot = np.std(bootstrap_means, ddof=1)
se_mu_boot
```

El error estandar de (b) es menor si no se usa bootstrap

#### (d)
- **Crear un intervalo de confianza del $95\%$ usando:** 
  $[\hat{\mu} - 2\text{SE}(\hat{\mu}),\; \hat{\mu} + 2\text{SE}(\hat{\mu})]$  
- **Comparar con el intervalo obtenido usando `Boston['medv'].std()`.**

```python
ci_lower = mu_hat - 2*se_mu_boot
ci_upper = mu_hat + 2*se_mu_boot
ci_lower, ci_upper
```

### 9.3. Mediana y percentil 10

#### (e)
- **Estimar la mediana $\hat{\mu}_{\text{med}}$ de `medv`.**

```python
mu_med_hat = np.median(boston['medv'])
mu_med_hat
```

#### (f)
- **Estimar su error estándar mediante bootstrap.**
- **Comentar hallazgos.**

```python
bootstrap_medians = np.empty(B)

for b in range(B):
    sample = rng.choice(boston['medv'], size=n, replace=True)
    bootstrap_medians[b] = np.median(sample)

se_median_boot = np.std(bootstrap_medians, ddof=1)
se_median_boot
```

La mediana de medv tiene un error estándar de ~0.38, lo que indica la variabilidad esperada de la mediana si tomáramos múltiples muestras de la población.

#### (g)
- **Estimar el percentil 10:**
  **$\hat{\mu}_{0.1} = \text{percentile}_{10}(\text{medv})$.**

```python
mu_01_hat = np.percentile(boston['medv'], 10)
mu_01_hat
```

#### (h)
- **Usar bootstrap para estimar el error estándar de $\hat{\mu}_{0.1}$.**
- **Comentar resultados.**

```python
bootstrap_percentiles = np.empty(B)

for b in range(B):
    sample = rng.choice(boston['medv'], size=n, replace=True)
    bootstrap_percentiles[b] = np.percentile(sample, 10)

se_percentile_boot = np.std(bootstrap_percentiles, ddof=1)
se_percentile_boot
```

El décimo percentil es más sensible a la variabilidad en los datos extremos, por lo que su error estándar es mayor que el de la media o incluso que la mediana.
