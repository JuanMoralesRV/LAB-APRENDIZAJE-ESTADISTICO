```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Advertising.csv", index_col="Unnamed: 0")

df = df.drop_duplicates().dropna()
```

```python
df.head()
```

### 1.- Tomar Advertising y hacer una regresión lineal como benchmark (R2)

```python
# Elegir columnas
num_columns = ['TV', 'radio', 'newspaper']

cat_columns = []

target = ["sales"]

# Separar datos
x = df[num_columns + cat_columns]
y = df[target]
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

numerical_transformer = StandardScaler() # Estandariza los datos

preprocessor = ColumnTransformer(
    transformers = [
        ("num", numerical_transformer, num_columns)
    ]
)
```

```python
from sklearn.linear_model import LinearRegression

modelo = LinearRegression()
```

```python
pipeline = Pipeline(steps = [
    ("preprocessor", preprocessor),
    ("model", modelo)
]   
)


pipeline.fit(x,y)
```

```python
from sklearn.metrics import r2_score

Y_1 = pipeline.predict(x)

r2_lineal_regression = r2_score(y,Y_1)

r2_lineal_regression
```

**2.- Usar un decision tree y ajustar con dataset (R2)**

```python
from sklearn.tree import DecisionTreeRegressor

modelo_tree = DecisionTreeRegressor()
```

```python
pipeline_tree = Pipeline(steps = [
    ("preprocessor", preprocessor),
    ("model", modelo_tree)
]   
)


pipeline_tree.fit(x,y)
```

```python
Y_2 = pipeline_tree.predict(x)

r2_tree = r2_score(y,Y_2)

r2_tree
```

**3.- Comparar R2. Escribe observaciones**

Con un regresion lineal me da como R2 0.89 y en el arbol me da R2 1

**4.- Repetir pasos 1 a 3 con train-test (random_state=42)**

```python
# Regresion Lineal traon - test


from sklearn.model_selection import train_test_split

# Dividir train y test
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)

# Crear el pipeline 
modelo_lineal = modelo 

pipeline_lineal = Pipeline(steps=[
    ("preprocessor", preprocessor),  # el preprocesador se ajustará solo con x_train
    ("model", modelo_lineal)
])

# Entrenar con train
pipeline_lineal.fit(x_train, y_train)

# Predecir sobre test
Y_pred_test = pipeline_lineal.predict(x_test)

# Calcular R2 en test
r2_lineal_regression_test = r2_score(y_test, Y_pred_test)

r2_lineal_regression_test
```

```python
# Arboles tree-test

# Regresion Lineal traon - test

from sklearn.model_selection import train_test_split

# Dividir train y test
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)

# Crear el pipeline 
modelo_tree_tt = DecisionTreeRegressor(random_state=42)

pipeline_tree_tt = Pipeline(steps=[
    ("preprocessor", preprocessor),  
    ("model", modelo_tree_tt)
])

# Entrenar con train
pipeline_tree_tt.fit(x_train, y_train)

# Predecir sobre test
Y_pred_test = pipeline_tree_tt.predict(x_test)

# Calcular R2 en test
r2_tree_tt = r2_score(y_test, Y_pred_test)

r2_tree_tt
```

**Observaciones**

La regresion mantiene casi el mismo R2 y el arbol sigue teniendo mejor metrico que la regresión, pero bajo su rendimiento

**5.- Repetir paso 1 a 4 con PolynomialFeatures(degree=2)**

```python
from sklearn.preprocessing import PolynomialFeatures

# Crear modelo poly
modelo_poly = LinearRegression()

# Crear pipeline
pipeline_poly = Pipeline(steps=[
    ("preprocessor", preprocessor),      
    ("poly_features", PolynomialFeatures(degree=2)),  
    ("model", modelo_poly)
])

# Entrenar con train
pipeline_poly.fit(x, y)

# Predecir sobre test
Y_pred_poly_test = pipeline_poly.predict(x)

# Calcular R² en test
r2_poly_test = r2_score(y, Y_pred_poly_test)

r2_poly_test
```

```python
from sklearn.preprocessing import PolynomialFeatures

# Crear modelo poly
modelo_poly_tree = DecisionTreeRegressor(random_state=42)

# Crear pipeline 
pipeline_poly_tree = Pipeline(steps=[
    ("preprocessor", preprocessor),          
    ("poly_features", PolynomialFeatures(degree=2)),  
    ("model", modelo_poly_tree)
])

# Entrenar con train
pipeline_poly_tree.fit(x, y)

# Predecir sobre test
Y_pred_poly_tree = pipeline_poly_tree.predict(x)

# Calcular R2 en test
r2_poly_test_tree = r2_score(y, Y_pred_poly_tree)

r2_poly_test_tree
```

**Observación**

SIn trai/test el modelo regresion lineal mejora con un R2 de 0.98 y el arbol igual sigue en R2 = 1
